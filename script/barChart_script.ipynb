```python
#imports necesarios para la visualización de data
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
path = Path("..") /"data" / "amazon.csv"
print(path) 

data_frame_amazon = pd.read_csv(path)
```

```python
productos_por_categoria = (
    data_frame_amazon["category"]
    .dropna()
    .str.split("|")
    .explode()
    .value_counts()
    .sort_values(ascending=False)
)

print(productos_por_categoria)
```

```python
#conversion de los productos a csv:
productos_por_categoria.to_csv
```
