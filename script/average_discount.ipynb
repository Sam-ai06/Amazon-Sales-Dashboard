```python
import csv
import json
from pathlib import Path

import pandas as pd
```

```python
project_root = Path.cwd().parent
data = pd.read_csv(project_root / 'data' / 'amazon.csv')
output_dir = project_root / 'public' / 'data_extracted'
```

```python
# El porcentaje llega como texto, por ejemplo: '64%'.
discounts = pd.to_numeric(data['discount_percentage'].str.rstrip('%'), errors='coerce')
average_discount = discounts.mean()
average_discount
```

```python
result = pd.DataFrame([{
    'title': 'Descuento promedio',
    'count': average_discount
}])

```

```python
csv_path = output_dir / 'average_discount.csv'
json_path = output_dir / 'average_discount.json'
result.to_csv(csv_path, index=False)

with csv_path.open('r', encoding='utf-8') as csv_file:
    records = list(csv.DictReader(csv_file))

with json_path.open('w', encoding='utf-8') as json_file:
    json.dump(records, json_file, indent=4)
```
