### Flujo Esperado

CSV original > Python/pandas > Data cleaning/agrupacion > CSV o JSON procesado > React > componente gráfico

```python
categories = set()

for category in amazon_products['category']:
    sub_categories = category.split("|")
    for sub_category in sub_categories:
        categories.add(sub_category)
 
print(sorted(categories))
print("=" * 60) 
print(f"numero de categorias identificadas en el dataset: {len(categories)}")
```

```python
productos_por_categoria = (
    dataframe_amazon_products['category']
    .dropna()
    .str.split("|")
    .explode()
    .value_counts()
    .sort_values(ascending=False)
)

print(productos_por_categoria)
```

```python
#conversión a csv para su uso en el componente de barras
productos_por_categoria.to_csv
```
