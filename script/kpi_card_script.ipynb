```python
# imports necesarios para la visualización y procesamiento de data
import pandas as pd
import numpy as np
from pathlib import Path
import csv
import json
```

```python
path = Path("..") / "data" / "amazon.csv"
print(path)

data_frame_amazon = pd.read_csv(path)
```

```python
# contenido del dataframe
print(f"Total de registros: {len(data_frame_amazon)}")
data_frame_amazon.info()
```

```python
# cálculo del total de productos para la métrica KPI
total_products = len(data_frame_amazon)
kpi_df = pd.DataFrame([{
    'title': 'Total de productos',
    'count': total_products
}])

print(kpi_df)
```

```python
# exportar métrica a CSV para el componente KPICardUI
kpi_df.to_csv('../public/data_extracted/total_products.csv', index=False)
```

```python
# convertir el CSV a JSON para consumo del frontend
with open('../public/data_extracted/total_products.csv', 'r', encoding='utf-8') as csv_file:
    data = list(csv.DictReader(csv_file))

with open('../public/data_extracted/total_products.json', 'w', encoding='utf-8') as json_file:
    json.dump(data, json_file, indent=4)
```
